```python
# all imports
import getpass
import os
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage
from langgraph.prebuilt import create_react_agent
```

```python
#set env parameters
if "TAVILY_API_KEY" not in os.environ:
  os.environ["TAVILY_API_KEY"] = getpass.getpass("Tavily API Key: ")

os.environ["LANGSMITH_TRACING"] = "true"
os.environ["LANGSMITH_PROJECT"] = "agenticai-langchain-demo"
if "LANGSMITH_API_KEY" not in os.environ:
  os.environ["LANGSMITH_API_KEY"] = getpass.getpass("LangSmith API Key: ")

if "GOOGLE_API_KEY" not in os.environ:
    #os.environ["GEMINI_API_KEY"] = getpass.getpass("Gemini API Key: ")
    os.environ["GOOGLE_API_KEY"] = getpass.getpass("Google API Key: ")
```

```python
#setup tools for the Agent

search = TavilySearchResults(max_results=2)
#search_results = search.invoke("What is the weather in Melbourne, Australia today?")
#print(search_results)
tools = [search]
```

```python
#Initiate the LLM
model = init_chat_model("gemini-2.5-pro-exp-03-25", model_provider="google_genai")
```

```python
#Create the Agent
agent_executor = create_react_agent(model, tools)

#Execute the agent
response = agent_executor.invoke({"messages": [HumanMessage(content="What's the weather in Melbourne, Australia like today?")]})
response["messages"]
```
